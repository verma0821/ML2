```python
import pandas as pd
from Orange.data import Table, Domain, DiscreteVariable
from Orange.classification.rules import CN2Learner

df = pd.read_csv("zoo.csv")

target_column = "class_type"
non_feature_columns = ["animal_name"] if "animal_name" in df.columns else []

X = df.drop(non_feature_columns + [target_column], axis=1)
y = df[target_column]

X = X.astype(str)
y = y.astype(str)

feature_vars = []

for col in X.columns:
    values = sorted(X[col].unique().tolist())
    feature_vars.append(DiscreteVariable(col, values=values))

target_values = sorted(y.unique().tolist())
target_var = DiscreteVariable(target_column, values=target_values)

domain = Domain(feature_vars, target_var)

data_orange = Table.from_list(
    domain,
    [list(row) + [label]
     for row, label in zip(X.values.tolist(), y.tolist())]
)

learner = CN2Learner()

classifier = learner(data_orange)

print("CN2 Classification Rules:")

for rule in classifier.rule_list:
    print(rule)
```
